# Occurrence Statistics Analysis

This notebook contains the essential post-processing analyses for the occurrence-based
crystal structure prediction study.

Included analyses:

1. Convergence of generated structures
2. Structural grouping statistics
3. Occurrence distribution analysis
4. Sampling-size sensitivity analysis
5. Occurrence-energy-experimental matching analysis

Non-essential exploratory plots and intermediate debugging cells have been removed.


## 1. Load packages and define paths

The notebook assumes that generated structures and grouping results are available
from the CSP workflow.


```
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tqdm import tqdm
from pymatgen.core import Composition
```

## 2. Check convergence of generated structures

Counts the number of successfully generated/relaxed structures and summarizes
the convergence behavior across compositions.


```
# Example: count generated structures per composition

species = [ii.split('_')[0] for ii in glob.glob('binary_final/cifs/*')]

num_gen = []
num_atoms = []

for sp in species:
    files = glob.glob(f'binary_final/cifs/{sp}/*.cif')
    num_gen.append(len(files))

print(f'Number of compositions: {len(species)}')
print(f'Average generated structures: {np.mean(num_gen):.2f}')
```

## 3. Structural grouping and occurrence calculation

Relaxed structures are grouped into structural basins.
Occurrence is calculated from the number of structures belonging to each basin.


```
# Load grouping information

group_files = glob.glob('binary_final/cifs/*/Groups_600*.csv')

print(f'Number of grouping files: {len(group_files)}')

group_sizes = []

for f in group_files:
    df = pd.read_csv(f)
    group_sizes.extend(df.iloc[:, -1].values)

print(f'Number of identified groups: {len(group_sizes)}')
```

## 4. Sampling-size sensitivity analysis

Comparison of occurrence statistics obtained from different numbers of sampled
structures (e.g. 200, 600, and 1000 structures).


```
# Compare grouping results from different sampling sizes

files_600 = glob.glob('binary_final/cifs/**/Groups_600*.csv', recursive=True)
files_1000 = glob.glob('binary_final/cifs/**/Groups_1000*.csv', recursive=True)

print('600 structures:', len(files_600))
print('1000 structures:', len(files_1000))
```
